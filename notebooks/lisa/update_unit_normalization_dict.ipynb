```python
import pandas as pd

# Dateipfade
xlsx_path_old = r"C:\Users\up19040\Projekte\LMU\Drittanbietervergleich\unit_normalization_dict_Laia_new.xlsx"
csv_path = r'C:\Users\up19040\Projekte\LMU\Programmieren mit LMU\information-extraction-pilot\data\normalization_units\unit_normalization_dict.csv'
xlsx_path = r'C:\Users\up19040\Projekte\LMU\Drittanbietervergleich\unit_mapping_dict20250819.xlsx'
csv_path_new = r'C:\Users\up19040\Projekte\LMU\Programmieren mit LMU\information-extraction-pilot\data\normalization_units\unit_normalization_dict_new.csv'

# CSV einlesen
df_csv = pd.read_csv(csv_path)

# Excel einlesen
df_xlsx = pd.read_excel(xlsx_path)

# Nur relevante Spalten auswählen und filtern
df_xlsx_filtered = df_xlsx[['unit_raw', 'group']].copy()

# 'kg' zu 'kg CO2e' umbenennen etc
df_xlsx_filtered.loc[df_xlsx_filtered['group'] == 'kg', 'group'] = 'kg CO2e'
df_xlsx_filtered.loc[df_xlsx_filtered['group'] == 'Mt', 'group'] = 'Mt CO2e'
df_xlsx_filtered.loc[df_xlsx_filtered['group'] == 'kt', 'group'] = 'kt CO2e'
df_xlsx_filtered.loc[df_xlsx_filtered['group'] == 't', 'group'] = 't CO2e'
# Spalten umbenennen, um zur CSV zu passen
df_xlsx_filtered = df_xlsx_filtered.rename(columns={'unit_raw': 'unit', 'group': 'normalized_unit'})

# Kombinieren und Duplikate entfernen
df_combined = pd.concat([df_csv, df_xlsx_filtered], ignore_index=True)
df_combined = df_combined.drop_duplicates(subset=['unit'], keep='first')

# Jetzt noch "factor" aus dem Sheet "standardization" mergen
df_standardization = pd.read_excel(xlsx_path_old, sheet_name="standardization")

# Merge via "normalized_unit"
df_combined = df_combined.drop(columns=["factor"], errors="ignore")
df_final = df_combined.merge(
    df_standardization[["normalized_unit", "factor"]],
    on="normalized_unit",
    how="left"
)

# Zurückschreiben
df_final.to_csv(csv_path_new , index=False)
```
